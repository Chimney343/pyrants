# UCT-sweep visualisation

Working notebook for the IS-MCTS UCT-sweep analysis. It loads the four completed
runs under `artifacts/ismcts/experiments/uct_sweep/`, aggregates the per-arm
metrics, and writes a self-contained interactive report plus a data digest into
`docs/uct-sweep-viz/`.

**Conclusion:** use `uct_c = 5.0` for the NN-training rollout (fallback band 4.0–5.0).

```python
import importlib.util
from pathlib import Path

GEN = Path("docs/uct-sweep-viz/generate_viz.py")
spec = importlib.util.spec_from_file_location("uct_viz", GEN)
uct_viz = importlib.util.module_from_spec(spec)
spec.loader.exec_module(uct_viz)

records, meta = uct_viz.load_sweep_data()
print(f"{len(records)} arm observations across {len(meta['runs'])} runs, "
      f"num_sims={meta['num_sims']}")
```

```python
# Per-arm table (win share, mean VP, mean rank, recruited deck entropy)
for r in records:
    print(f"uct={r['uct']:>4g}  win={100*r['win_rate']:5.1f}%  "
          f"VP={r['mean_vp']:6.2f}  rank={r['mean_rank']:.3f}  "
          f"entropy={r['entropy_mean']:.3f}  ({r['label']})")
```

```python
figures = uct_viz.build_figures(records)
print("figures:", list(figures))

# Write the self-contained HTML report + data.json into docs/uct-sweep-viz/
uct_viz.write_outputs()
```
